```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Training data
# Feature matrix with a single feature "House size" measured in sq. ft.
X = np.array([
    [1000],
    [1500],
    [2000],
    [2500],
    [3000]
])

# Target Vector
# Price of house
y = np.array([
    200000,
    275000,
    350000,
    425000,
    500000
])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Make a prediction
prediction = model.predict([[1800]])

print(prediction)
```

```python
# Retrieve learned parameters
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (w): {slope}")
print(f"Intercept (b): {intercept}")

# Display the learned equation
print(f"Learned formula: y = {slope:.2f}x + {intercept:.2f}")

print(f"Predicted house price for a house of 1800 sq ft: ${prediction[0]:,.2f}")
```

```python
import matplotlib.pyplot as plt

# Generate points for the regression line
X_line = np.linspace(900, 3100, 100).reshape(-1, 1)
y_line = model.predict(X_line)

# Plot training data
plt.scatter(X, y, color="blue", label="Training Data")

# Plot regression line
plt.plot(X_line, y_line, color="red", label="Regression Line")

# Plot prediction
plt.scatter(
    1800,
    prediction[0],
    color="green",
    s=120,
    marker="*",
    label=f"Prediction: ${prediction[0]:,.0f}",
    zorder=5
)

# Labels and formatting
plt.xlabel("House Size (sq. ft.)")
plt.ylabel("House Price ($)")
plt.title("Linear Regression: House Price Prediction")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
