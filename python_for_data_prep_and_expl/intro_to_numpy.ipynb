```python
import numpy as np

# Python list
numbers = [1,2,3,4,5]

# NumPy array
arr = np.array([1,2,3,4,5])

print(numbers)
print(arr)
print(arr.mean())
print(arr.sum())
print(arr.std())

sum = 0
for item in numbers:
    sum += item
print(sum/len(numbers))
```

```python
# Create a two-dimensional array
data = np.array([
    [22, 45000],
    [25, 52000],
    [30, 60000],
    [35, 75000],
    [40, 85000]
])

print(data)
```

```python
# Array Attributes

# Number of dimensions
print(arr.ndim)
print(data.ndim)

# Shape of the array (i.e., number of rows and columns)
print(arr.shape)
print(data.shape)

# Total number of elements
print(arr.size)
print(data.size)

# Data type of the elements of the array
print(arr.dtype)
print(data.dtype)
```

```python
# Creating Arrays with Specific Shapes

# Array of zeros
zeros = np.zeros((3, 2))

# Array of ones
ones = np.ones((2, 3))

# Sequence of numbers
numbers = np.arange(0, 10, 2)

# Equally spaced values
values = np.linspace(0, 1, 5)

print(zeros)
print(ones)
print(numbers)
print(values)
```

```python
# Indexing and modifying arrays

ages = np.array([22, 25, 30, 35, 40])

# Access individual elements
print(ages[0])
print(ages[2])
print(ages[-1])

# Modify an element
ages[0] = 23

print(ages)
```

```python
# Indexing a 2D array (list is shown in the next code cell)

data = np.array([
    [22, 45000],
    [25, 52000],
    [30, 60000],
    [35, 75000],
    [40, 85000]
])

# Access the first row
print(data[0])

# Access the first column of the first row
print(data[0, 0])

# Access the second column of the third row
print(data[2, 1])
print(data[2][1])
```

```python
# Indexing a 2D list

data = [
    [22, 45000],
    [25, 52000],
    [30, 60000],
    [35, 75000],
    [40, 85000]
]

# Access the first row
print(data[0])

# Access the first column of the first row
print(data[0][0])

# Access the second column of the third row
print(data[2][1])
print(data[2, 1])
```

```python
# Array Slicing

ages = np.array([22, 25, 30, 35, 40, 45, 50])

# First three elements
print(ages[:3])

# Elements from index 2 to 4
print(ages[2:5])

# Every second element
print(ages[::2])

# Last three elements
print(ages[-3:])
```

```python
# 2D Array Slicing

data = np.array([
    [22, 45000],
    [25, 52000],
    [30, 60000],
    [35, 75000],
    [40, 85000]
])

# First three rows
print(data[:3, :])

# All rows, first column
print(data[:, 0])

# All rows, second column
print(data[:, 1])
```

```python
# Vectorized Operations

# - NumPy Allows us to perform operations on entire arrays 
#   without explicitly writing loops which is one of its biggest
#   advantages

# Traditional Python Approach

numbers = [1, 2, 3, 4, 5]

result = []

for number in numbers:
    result.append(number * 2)

print(result)

# Numpy Approach

numbers = np.array([1, 2, 3, 4, 5])

result = numbers * 2

print(result)
```

```python
# More vectorized operations

numbers = np.array([10, 20, 30, 40, 50])

print(numbers + 5)
print(numbers - 5)
print(numbers * 2)
print(numbers / 10)
print(numbers ** 2)
```

```python
# Operations between two arrays

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)
print(a * b)
```

```python
# Operations between two lists
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)
print(a * b)
```

```python
# Statistical and aggregation functions

ages = np.array([22, 25, 30, 35, 40, 45, 50])

# Mean
print(np.mean(ages))

# Median
print(np.median(ages))

# Standard deviation
print(np.std(ages))

# Minimum and maximum
print(np.min(ages))
print(np.max(ages))

# Sum
print(np.sum(ages))

# Variance
print(np.var(ages))
```

```python
# Statisical and aggregate functions across dimensions

data = np.array([
    [22, 45000],
    [25, 52000],
    [30, 60000],
    [35, 75000],
    [40, 85000]
])

# Average of each column
print(np.mean(data, axis=0))

# Average of each row
print(np.mean(data, axis=1))
```

```python
import numpy as np

# Features: House size (sq. ft.)
X = np.array([
    [1000],
    [1500],
    [2000],
    [2500],
    [3000]
])

# Labels: House prices ($)
y = np.array([
    200000,
    275000,
    350000,
    425000,
    500000
])

print("Features (X):")
print(X)

print("\nLabels (y):")
print(y)

print("\nFeature shape:", X.shape)
print("Label shape:", y.shape)
```

### Why is X two-dimensional while y is one-dimensional?

Scikit-learn estimators expect the feature matrix to have the shape (n_samples, n_features), whereas the target for a standard single-output regression problem is one-dimensional (n_samples)
