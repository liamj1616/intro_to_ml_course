````python
# Identify missing values

df.isnull().sum()
````

### Options for dealing with missing values

1. Remove rows with missing values

```python
df.dropna()
```

2. Replace missing numerical values

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)
```

Common imputation approaches:
- Median imputation (as shown above)
- Mean imputation
- Mode imputation
- Predictive imputation

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram showing Distribution of Ages
sns.histplot(df["Age"], kde=True)
plt.show()

# Box plot showing Distribution of Ages
sns.boxplot(x=df["Age"])
plt.show()

# Correlation heatmap
sns.heatmap(
    df.select_dtypes("number").corr(),
    annot=True,
    cmap="coolwarm"
)
plt.show()
````

````python
# Calculate average survival rate by passenger class
survival_by_class = df.groupby("Pclass")["Survived"].mean()

print(survival_by_class)

# Create bar plot
sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    errorbar=None
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.show()
````

````python
# Calculate average survival rate by sex
survival_by_sex = df.groupby("Sex")["Survived"].mean()

print(survival_by_sex)

# Create bar plot
sns.barplot(
    data=df,
    x="Sex",
    y="Survived",
    errorbar=None
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.show()
````

````python
# Calculate Average Survival Rate by Sex and Class

sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    hue="Sex",
    errorbar=None
)

plt.title("Survival Rate by Passenger Class and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)
plt.legend(title="Sex")

plt.show()
````
